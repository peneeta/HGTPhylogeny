# Plot the intragenic conflict results

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import seaborn as sns
```

Plotting Function

```python
def PlotIntragenicConflictHeatmap(df, title):
    colors = ["firebrick", "white", "dodgerblue"]
    cmap = ListedColormap(colors)

    labels = ["conflict", "no tree constructed", "concordant"]
    values = [-1, 0, 1]

    # Plot
    sns.heatmap(df.T,
                cmap=cmap,
                cbar_kws={'ticks': values},
                linewidths=0.4,          # Thickness of the borders
                linecolor='black')

    # Update colorbar labels
    colorbar = plt.gca().collections[0].colorbar
    colorbar.set_ticklabels(labels)

    plt.ylabel("Gene Segment")
    plt.title(f"Intragenic Conflict Heatmap for {title}")

    plt.xticks([], [])

    plt.xlabel("Clade")

    # Show the plot
    plt.show()
```

```python
rrna_df = pd.read_csv("./16S/subset_16S_aligned_trimmed.fasta.interval_plotdata", sep=" ")
gyra_df = pd.read_csv("./gyrA/subset_gyrA_aligned_trimmed.fasta.interval_plotdata", sep=" ")
```

```python
PlotIntragenicConflictHeatmap(rrna_df, "16S rRNA")
```

```python
PlotIntragenicConflictHeatmap(gyra_df, "gyrA")
```
