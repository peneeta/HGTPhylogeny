# Plot the intragenic conflict results

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import seaborn as sns
```

```python
def PlotIntragenicConflictHeatmap(df, title):
    mapping = {-1: "conflict",
               0: "unresolved", 
               1: "concordant"}
    
    # All possible values in order
    all_values = [-1, 0, 1]
    all_colors = ["firebrick", "white", "dodgerblue"]
    
    # Filter to only values present in df
    present_values = [v for v in all_values if v in df.values]
    present_colors = [all_colors[all_values.index(v)] for v in present_values]
    present_labels = [mapping[v] for v in present_values]

    # Create colormap for only present values
    cmap = ListedColormap(present_colors)

    # Plot
    sns.heatmap(df,
                cmap=cmap,
                cbar_kws={'ticks': present_values},
                linewidths=0.4,
                linecolor='black')

    # Update colorbar labels
    colorbar = plt.gca().collections[0].colorbar
    colorbar.set_ticklabels(present_labels)

    plt.ylabel("Clade")
    plt.title(f"Intragenic Conflict Heatmap for {title}")
    plt.yticks([], [])
    plt.xlabel("Gene Segment")
    plt.show()
```

## Diverse Species List

```python
rrna_df = pd.read_csv("./diverse/16S/subset_16S_aligned_trimmed.fasta.interval_plotdata", sep=" ")
PlotIntragenicConflictHeatmap(rrna_df, "16S rRNA (Diverse)")
```
